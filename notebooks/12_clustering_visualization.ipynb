```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../datasets/master/master_final_2025.csv')
```

```python
# Correct, honest cluster labels based on actual profile differences (not forced quadrant naming)
cluster_names = {
    0: 'Cluster 0: Mayoritas/Baseline (n=26)',
    1: 'Cluster 1: Krisis Ekstrem (n=2)',
    2: 'Cluster 2: Akses Sekunder Tinggi/Urban (n=5)',
    3: 'Cluster 3: Kaya Perpustakaan (n=5)',
}
df['cluster_label'] = df['kmeans_cluster'].map(cluster_names)
df.to_csv('../datasets/master/master_final_2025.csv', index=False)

fig, ax = plt.subplots(figsize=(11,9))
palette = {0:'#2E7D32', 1:'#B71C1C', 2:'#1565C0', 3:'#8E24AA'}
for c in sorted(df['kmeans_cluster'].unique()):
    sub = df[df['kmeans_cluster']==c]
    ax.scatter(sub['pc1'], sub['pc2'], s=100, c=palette.get(c,'grey'), label=cluster_names[c], edgecolor='white', linewidth=0.8, zorder=3)
for _, r in df.iterrows():
    ax.annotate(r['province'].title(), (r['pc1'],r['pc2']), fontsize=7.5, xytext=(4,4), textcoords='offset points', alpha=0.85)
ax.set_xlabel('PC1 (41% variance)')
ax.set_ylabel('PC2 (23% variance)')
ax.set_title('KMeans Clustering (k=4, data-driven, 8 fitur, RobustScaler)\nSilhouette score = 0.33 — separasi moderat', fontsize=12, fontweight='bold')
ax.legend(loc='upper left', fontsize=9)
plt.tight_layout()
plt.savefig('../src/visualization/07_kmeans_pca.png', dpi=150)
plt.close()
print("done")
```
